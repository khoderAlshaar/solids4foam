```python
# ------------------------------------------------------------
# Compute total (stagnation) pressure and temperature
# from static (ambient) conditions and velocity.
# ------------------------------------------------------------

import math

# --- Input: ambient (static) conditions ---
p_static = 1e5    # Pa (ambient pressure)
T_static = 288.15       # K (ambient temperature)
U = 1            # m/s (flow velocity)
# Set throuhg the Stiffned Gas Equation of state 
gamma = 2.782917482129208    
p_inf =  773221168.8164893
Cv = 1505.254836659763 # specific heat ratio for Stiffened Gas
R = (gamma-1) * Cv              # Stiffened gas constant for water [J/kg/K]

# --- Derived quantities ---
a = math.sqrt(gamma * R * T_static)        # speed of sound
M = U / a                                  # Mach number

T_total = T_static * (1 + (gamma - 1) / 2 * M**2)

p_total  =( (p_static + p_inf) * (1 + (gamma - 1) / 2 * M**2)**(gamma / (gamma - 1))) - p_inf

# --- Output ---
print("--------------------------------------------------")
print(" Isentropic Relations: Ambient → Total Conditions")
print("--------------------------------------------------")
print(f"Input (Ambient Conditions):")
print(f"  Static pressure (p)     : {p_static:.3f} Pa")
print(f"  Static temperature (T)  : {T_static:.3f} K")
print(f"  Velocity (U)            : {U:.6f} m/s\n")

print("Computed quantities:")
print(f"  Speed of sound (a)      : {a:.3f} m/s")
print(f"  Mach number (M)         : {M:.8f}")
print(f"  Total temperature (T0)  : {T_total:.6f} K")
print(f"  Total pressure (p0)     : {p_total:.6f} Pa")
print("--------------------------------------------------")
```
